# Post 2: a d20 GPT on ClimbMix, on a Colab accelerator

Trains a 20-block GPT on the web text nanochat uses, with the tokenizer from
`tokenizer/bpe.py`. Text is tokenized once, ahead of training.

Checkpoints and token files go straight to a GCS bucket and metrics go to Weights &
Biases, so all of them outlive the runtime. **If the runtime goes away, start a new one
and run the notebook again: it reuses the token files and carries on from the newest
checkpoint.**

`COLAB.md` explains how to run this with `colab exec`, which runs every cell in order with
nobody at the keyboard. With `--env SMOKE=1`, the same notebook runs a tiny model on two
shards, which proves every step in minutes on any runtime.

## The run

| Setting | `local` | This run |
| --- | --- | --- |
| Data | TinyStories | ClimbMix, 120 shards (about 6.4B tokens) |
| Blocks / dim / heads | 4 / 256 / 8 | 20 / 1,280 / 10 |
| Vocabulary | 8,192 | 32,768, trained on 8 shards |
| Context | 256 | 2,048 |
| Parameters | 7.3M | 477M |
| Batch | 32 | 64, as 4 microbatches of 16 |
| Tokens seen | 41M | 5.7B, 12 per parameter |
| Precision | float32 | bfloat16 for compute, float32 for parameters |

The shape, batch and context come from the earlier A100 run of this model, and 12 tokens
per parameter is nanochat's default. Learning rates are the ones `local` trained well on:
Muon at 0.02 on the blocks, and Adam at 0.2 and 0.004 on the embeddings and output layer,
scaled by the square root of 768 / 1,280, with warmup and cosine decay. nanochat's defaults
have since moved to 0.3 and 0.008, so those are the next thing to try.

Gradient checkpointing (remat) trades speed for memory, and is on. If this runs out of
memory, halve `BATCH_PER_DEVICE` and double `NUM_MICROBATCHES`.

```python
import jax
import jax.numpy as jnp
import numpy as np
from absl import logging as absl_logging

from toylib.harness import analyze
from toylib.harness import metrics as harness_metrics
from toylib_projects.p02_tinystories import data, experiment, tinystories, train, webtext
from toylib_projects.p02_tinystories import metrics as metrics_module

DEPTH = 20
SEQ_LEN = 2048
VOCAB_SIZE = 32_768
BATCH_PER_DEVICE = 16
NUM_MICROBATCHES = 4
NUM_TRAIN_SHARDS = 120  # About 53M tokens each.
NUM_TOKENIZER_SHARDS = 8
MAX_STEPS = 44_000  # 12 tokens per parameter, at 131,072 tokens a step.
WARMUP_STEPS = 500
MUON_LR, EMBED_LR, OUTPUT_LR = 2e-2, 2e-1, 4e-3
MAX_GRAD_NORM = 1.0
EVAL_INTERVAL_STEPS = 500
NUM_EVAL_STEPS = 20
SAVE_INTERVAL_STEPS = 2_500  # The most work a lost runtime can cost.
MAX_TO_KEEP = 3  # A checkpoint is several GB, with optimizer state.
DTYPE = jnp.bfloat16
TIMING_STEPS = 20
RUN_ID = f"d{DEPTH}_v{VOCAB_SIZE}"

if SMOKE:
    DEPTH, SEQ_LEN, VOCAB_SIZE, BATCH_PER_DEVICE, NUM_MICROBATCHES = 2, 64, 1_000, 2, 2
    NUM_TRAIN_SHARDS, NUM_TOKENIZER_SHARDS = 2, 2
    MAX_STEPS, WARMUP_STEPS, EVAL_INTERVAL_STEPS, NUM_EVAL_STEPS = 21, 5, 10, 2
    SAVE_INTERVAL_STEPS, TIMING_STEPS, DTYPE = 10, 3, jnp.float32
    RUN_ID = "smoke"

ROOT = pathlib.Path("/content" if IN_COLAB else train.OUTPUT_DIR)
DATA_DIR = pathlib.Path(os.environ.get("DATA_DIR", ROOT / "climbmix"))
CHECKPOINT_ROOT = f"{BUCKET}/toylib-p02-climbmix" if IN_COLAB else str(ROOT / "climbmix_runs")
DATA_BUCKET = f"{BUCKET}/toylib-p02-climbmix/data_shards{NUM_TRAIN_SHARDS}"

absl_logging.set_verbosity(absl_logging.ERROR)  # quiet orbax's read-only notices
print("JAX", jax.__version__, "| devices:", jax.devices())
for device in jax.local_devices():
    limit = (device.memory_stats() or {}).get("bytes_limit")
    print(f"{device.device_kind}: {limit / 2**30:.1f} GiB" if limit else device.device_kind)
```

## Tokens

The first run downloads the shards, trains the tokenizer, writes the token files and copies
them to the bucket. Later runs only copy them back. Measured on a laptop, a shard takes
about 35 seconds on one core, and training the tokenizer a few minutes.

This needs no accelerator. To avoid paying for one during this step, run the notebook once
on a CPU runtime: it stops after this cell, with the token files in the bucket.

```python
token_files = [
    tinystories.tokenizer_path(VOCAB_SIZE, DATA_DIR),
    tinystories.bin_path("train", VOCAB_SIZE, DATA_DIR),
    tinystories.bin_path("val", VOCAB_SIZE, DATA_DIR),
]
DATA_DIR.mkdir(parents=True, exist_ok=True)
if IN_COLAB:
    in_bucket = !gcloud storage ls {DATA_BUCKET}/ 2>/dev/null
    for path in token_files:
        if not path.exists() and f"{DATA_BUCKET}/{path.name}" in in_bucket:
            !gcloud storage cp {DATA_BUCKET}/{path.name} {path}

webtext.prepare(VOCAB_SIZE, NUM_TRAIN_SHARDS, DATA_DIR, num_tokenizer_shards=NUM_TOKENIZER_SHARDS)

if IN_COLAB:
    for path in token_files:
        if f"{DATA_BUCKET}/{path.name}" not in in_bucket:
            !gcloud storage cp {path} {DATA_BUCKET}/{path.name}

tokenizer = tinystories.load_tokenizer(VOCAB_SIZE, DATA_DIR)
train_tokens = token_files[1].stat().st_size // 2
print(f"{len(tokenizer):,} tokenizer tokens, {train_tokens:,} training tokens, "
      f"{token_files[2].stat().st_size // 2:,} validation tokens")

if IN_COLAB and not SMOKE and jax.default_backend() == "cpu":
    raise RuntimeError("Token files are in the bucket. Training needs an accelerator; see COLAB.md.")
```

```python
def make_experiment(run_id=RUN_ID, logger_config=None, checkpoint_dir=None):
    """One run through the harness. Defaults to the full run, on W&B and GCS."""
    run_dir = ROOT / "climbmix_logs" / run_id
    run_dir.mkdir(parents=True, exist_ok=True)
    bpt_path = run_dir / "bytes_per_token.npy"
    np.save(bpt_path, tokenizer.bytes_per_token())

    batch_size = BATCH_PER_DEVICE * jax.local_device_count() * NUM_MICROBATCHES
    def dataset(split, seed, repeatable):
        return data.TokenBinDataset(
            path=str(tinystories.bin_path(split, VOCAB_SIZE, DATA_DIR)),
            batch_size=batch_size,
            seq_len=SEQ_LEN,
            seed=seed,
            repeatable=repeatable,
            tokenizer=tokenizer,
        )

    def schedule(peak_lr):
        return train.warmup_cosine_schedule(peak_lr, MAX_STEPS, WARMUP_STEPS)

    model_config = train.get_model_config(DEPTH, SEQ_LEN, VOCAB_SIZE, dtype=DTYPE)
    dmodel_lr_scale = (model_config.qkv_dim / 768) ** -0.5
    return experiment.Experiment(
        model_config=model_config,
        training_config=experiment.TrainingConfig(
            max_steps=MAX_STEPS,
            num_microbatches=NUM_MICROBATCHES,
            max_grad_norm=MAX_GRAD_NORM,
            optimizer_config=train.create_muon_adam_multi_optimizer_config(
                muon_lr=schedule(MUON_LR),
                adamw_embed_lr=schedule(EMBED_LR * dmodel_lr_scale),
                adamw_output_lr=schedule(OUTPUT_LR * dmodel_lr_scale),
            ),
        ),
        eval_config=experiment.EvalConfig(
            eval_interval_steps=EVAL_INTERVAL_STEPS, num_eval_steps=NUM_EVAL_STEPS
        ),
        checkpoint_config=experiment.CheckpointConfig(
            save_interval_steps=SAVE_INTERVAL_STEPS,
            save_final=True,
            max_to_keep=MAX_TO_KEEP,
            checkpoint_dir=checkpoint_dir or f"{CHECKPOINT_ROOT}/{run_id}",
            checkpoint_dataset_iterator=True,  # Resume with the same batches.
        ),
        logger_config=logger_config
        or (experiment.WandBLoggerConfig(
            project_name=WANDB_PROJECT,
            user_name=WANDB_ENTITY,
            log_dir=str(run_dir),
            run_id=run_id,  # A repeat of this run resumes the same W&B run.
            train_log_interval_steps=10,
            also_log_to_file=True,
        ) if IN_COLAB else experiment.LoggerConfig(
            log_dir=str(run_dir), run_id=run_id, train_log_interval_steps=10
        )),
        train_task=experiment.Task("train", dataset("train", seed=0, repeatable=False)),
        eval_task=experiment.Task(
            "val",
            dataset("val", seed=1, repeatable=True),
            metrics=[harness_metrics.Loss(), metrics_module.BitsPerByte(str(bpt_path))],
        ),
        sampling_max_tokens=40,
    )
```

## How long will this take?

A short timing run, saved nowhere. It also proves that the model fits and that training
and evaluation work on this accelerator before hours go into them. Peak memory is printed
where the accelerator reports it. An out-of-memory error here is the cue to change the
batch settings above.

```python
import tempfile
import time

with tempfile.TemporaryDirectory() as tmp:
    tmp = pathlib.Path(tmp)
    exp = make_experiment(
        "timing",
        logger_config=experiment.LoggerConfig(log_dir=str(tmp), run_id="timing"),
        checkpoint_dir=str(tmp / "checkpoints"),
    )
    exp.init_state()
    batches = iter(exp.train_task.dataset)
    exp.inner_loop(next(batches))  # Compiles the training step.
    jax.block_until_ready(exp.model)

    start = time.monotonic()
    for _ in range(TIMING_STEPS):
        exp.inner_loop(next(batches))
    jax.block_until_ready(exp.model)
    seconds_per_step = (time.monotonic() - start) / TIMING_STEPS

    exp.run_validation()  # Compiles the evaluation step.
    start = time.monotonic()
    exp.run_validation()
    seconds_per_eval = time.monotonic() - start

    n_params = analyze.print_param_sizes(exp.model, depth=1)[1]
    peak = (jax.local_devices()[0].memory_stats() or {}).get("peak_bytes_in_use")
    exp.cleanup()

num_evals = MAX_STEPS // EVAL_INTERVAL_STEPS
hours = (MAX_STEPS * seconds_per_step + num_evals * seconds_per_eval) / 3600
tokens = MAX_STEPS * BATCH_PER_DEVICE * jax.local_device_count() * NUM_MICROBATCHES * SEQ_LEN
print(f"{n_params:,} params  {seconds_per_step:.2f} s/step  {seconds_per_eval:.1f} s/eval  "
      f"about {hours:.1f} h for {MAX_STEPS:,} steps, plus checkpoints")
print(f"{tokens:,} tokens, {tokens / n_params:.1f} per parameter, "
      f"{tokens / train_tokens:.2f} passes over the data")
if peak:
    print(f"peak memory {peak / 2**30:.1f} GiB")
```

## Train

Prints a W&B link, where the curves and samples appear as it trains. It starts from the
newest checkpoint in the bucket if there is one.

```python
exp = make_experiment()
try:
    exp.init_state()
    resumed = exp.resume()
    print(("resuming at step " + str(exp.step)) if resumed else "starting fresh", f"of {MAX_STEPS:,}")
    exp.outer_loop()
    final = exp.run_validation()
    exp.sampling_evaluation()
finally:
    exp.cleanup()  # Waits for W&B to finish syncing.
print(final)
```
